```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

```

```python
df = pd.read_csv('3.17_2.74.txt', sep=',')
df["RUNTIME (MS)"] = df["RUNTIME (MS)"].sub(df["RUNTIME (MS)"].iloc[0])
df["RUNTIME (MS)"] = df["RUNTIME (MS)"].divide(60_000)


# clean column names and plot
df.columns = df.columns.str.strip().str.replace('\ufeff', '')
y_col = 'VOLTAGE (V)'
if y_col not in df.columns:
    matches = [c for c in df.columns if 'VOLTAGE' in c.upper()]
    if matches:
        y_col = matches[0]
    else:
        raise KeyError(f"Could not find a 'VOLTAGE' column. Available columns: {list(df.columns)}")

df.plot(x='RUNTIME (MS)', y=y_col)
plt.xlabel('RUNTIME (MS)')
plt.ylabel(y_col)
plt.tight_layout()
f = plt.figure()
f.set_figwidth(4)
f.set_figheight(1)
plt.show()
```
